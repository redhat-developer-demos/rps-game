# Request Model

This notebook demonstrates how to **send inference requests** to a deployed YOLO model using a REST API, typically served via **KServe**.  

It shows how to format an image input, send it to the model, and visualize the output, completing the final step of the ML workflow: **inference in production**.


* We configure the inference endpoint by specifying the URL and model name exposed via KServe.  
* We also define the local path to the test image that will be used for inference.

```python
import requests
import json
import numpy as np
import cv2
import matplotlib.pyplot as plt
import os
from remote_infer import process_image

KSERVE_URL = "REPLACE_WITH_MODEL_ENDPOINT_URL"
MODEL_NAME = "REPLACE_WITH_MODEL_ENDPOINT_NAME"

# Define KServe inference endpoint
MODEL_URL = f"{KSERVE_URL}/v2/models/{MODEL_NAME}/infer"

print(MODEL_URL)

# Define the local image path
HOME = os.getcwd()
destination_dir = os.path.join(HOME, "datasets", "data")  # Root directory
image_name = "20220216_221550_jpg.rf.02a071a383151953fcf8671fc7fca3af.jpg"
image_path = os.path.join(destination_dir, "test", "images", image_name)
print(image_path)
```

The image is processed and sent to the KServe endpoint using a helper function (`process_image`).  
The model returns annotated results (bounding boxes, class labels), which are rendered using `matplotlib` for visual validation.

```python
# Perform inference
img = process_image(image_path, MODEL_URL)

# Convert BGR to RGB for visualization
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Display the result
plt.figure(figsize=(8, 6))
plt.axis("off")
plt.imshow(img)
plt.show()
```
